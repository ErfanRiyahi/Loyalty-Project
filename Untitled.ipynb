```python
import pandas as pd
import numpy as np
```

```python
co = pd.read_csv(r'data/customer order.csv')
```

```python
du = pd.read_csv(r'data/DataCDR.csv')
```

```python
vu = pd.read_csv(r'data/voicesmsCDR.txt')
```

```python
from loyalty.standards.utils import clean_data
co_cleaned = clean_data(co, ['MSISDN', 'PaidDateUtc', 'OrderTotal'])
```

```python
du_cleaned = clean_data(du, ['SERVED_MSISDN', 'START_TIMESTAMP', 'ACTUAL_USAGE'])
```

```python
vu_cleaned = clean_data(vu, ['Subscriber_MSISDN', 'Call_Date', 'Call_Duration'])
```

```python
def clean_data(df: pd.Series, columns):
    user_id, date_name, usage_name = columns

    # Remove all nan items
    df = df.dropna().copy()

    df[date_name] = pd.to_datetime(df[date_name])
    df[usage_name] = pd.to_numeric(df[usage_name])

    # Convert date to year-month e.g. 2025-01
    df["YearMonth"] = df[date_name].dt.to_period("M")

    from loyalty.identifiers import normalize_identifiers
    df['MSISDN'] = normalize_identifiers(df[user_id])
    df = df.dropna(subset=['MSISDN']).copy()

    if "OrderStatusId" in columns:
        # Select users with successful order
        df = df[df["OrderStatusId"] == 30]
    return df
```
